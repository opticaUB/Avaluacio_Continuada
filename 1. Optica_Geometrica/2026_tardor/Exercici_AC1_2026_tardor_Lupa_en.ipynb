## **Exercise 1** of Continuous Assessment: Optical Instruments

### The mobile phone as an eye: an experimental study of the magnifying glass

<br>

Write your **full name**.

<br>

From an optical point of view, a mobile phone camera is very similar to the eye: it has a converging lens system (the objective, which plays the role of the crystalline lens), a diaphragm (the pupil), a sensor where the image is formed (the retina) and an autofocus system (accommodation). In this exercise you will use your phone as an **artificial eye** to study the **magnifying glass** experimentally: what angular magnification it provides, what portion of the object it lets you see (the field) and how the illumination is distributed in the image, depending on the distance between the magnifying glass and the eye.

The exercise has two parts: a small **experiment** that you will carry out with your phone and a magnifying glass, and the **theoretical development** needed to interpret the experiment.

#### Material

- Your mobile phone (use the main rear camera).
- A magnifying glass, i.e. a converging lens, with a focal length between approximately 5 and 20 cm. Any magnifying glass from a bazaar or stationery shop will do (if you cannot find one, ask the lecturer).
- Graph paper with a millimetre grid (you can print the one attached to the task on the Virtual Campus).
- A ruler or a measuring tape.
- Optional: a piece of black cardboard to better define the edge of the magnifying glass.
- Optional: books or boxes to use as supports.

#### Preparing the phone and the set-up

**Camera settings**

- Always use the main camera (1x), without digital zoom and without flash. If your phone has an automatic macro mode, turn it off: many phones switch camera without warning when the object is close.
- In the last photos the focus must be **locked at infinity**, like a relaxed eye that does not accommodate.
    - on an iPhone, focus on a distant object and lock the focus by keeping your finger pressed on the screen until «AE/AF Lock» appears.
    - on Android, use the Pro or Manual mode (MF focus at ∞) or a camera app with manual focus (for example, Open Camera).
- Each photograph stores metadata that includes, among other things, the actual focal length of the objective $f'_c$, the f-number $N$, the 35 mm equivalent focal length and the camera used. You can check them in the phone's gallery (photo details or information) or on a computer (on Windows: Properties → Details).

**Edge of the magnifying glass**

If the frame is not sharply enough defined in the last photo, you can stick on it a piece of black cardboard with a centred circular hole of known diameter, slightly smaller than the magnifying glass. The effective diameter of the magnifying glass $D_L$ will then be that of the hole.

**Suggested set-up**

A vertical set-up is usually the most convenient: the graph paper on the table and the magnifying glass parallel to the paper, held at the appropriate height with a pile of books. This keeps the magnifying glass steady and leaves your hands free. The phone goes on top, parallel to the magnifying glass and with the camera centred, held by another pile of books (or by hand) so that you can vary the distance $e$ between the magnifying glass and the phone. If it is too dark, make sure that light reaches the paper from the side, so that neither the phone nor the magnifying glass casts a shadow on it (never with the camera flash). Take a photo of the set-up to include it in your report.

<br>

#### Guide to the exercise

**1. Focal length of the magnifying glass.** Stand in front of a wall that has a window on the opposite wall. Bring the magnifying glass closer to the wall until a sharp image of the window is formed on it (if the image of the outside is formed, even better). Estimate the focal length $f'_L$ of the magnifying glass from the magnifying glass–wall distance. Take a photo showing the magnifying glass and the image of the window on the wall.

**2. Near point and reference photo.** Without the magnifying glass, bring your eye closer to the graph paper until you find your near point and measure its distance $s_{pp}$. Place the phone at the same distance from the paper and take a photo in which the graph paper is framed and in focus, with autofocus on (normal mode). What is the field of view of the camera (count the squares of the paper that fit in the photograph)? Using the metadata, deduce the size of the sensor and the diameter of the aperture stop of the camera $D_c$.

**3. Magnifying glass against the camera ($e = 0$).** Set the camera to focus at infinity and lock this setting. Place the magnifying glass centred in front of the camera, as close as you can. Find the magnifying glass–paper distance at which the paper is in focus and take a photo.
- What is the distance between the paper and the magnifying glass? Why?
- Comparing this photo with the one from step 2, obtain the angular magnification provided by the magnifying glass and compare it with its nominal value. Justify your answer.
- What is the field of view now? What limits it?

**4. Moving the camera away from the magnifying glass ($e > 0$).** Keeping the paper–magnifying glass distance fixed and the camera focus at infinity, move the camera away from the magnifying glass until its whole frame can be seen in the photo. Take a photo and measure the distance $e$ between the camera and the magnifying glass.
- Is the graph paper still in focus?
- What is the magnification now?
- What is the field of view? What limits it now?
- Measure the radius of the region with full illumination and that of the region with limiting illumination by counting the small squares of the graph paper.

**5. On-axis ray tracing.** For the case $e > 0$, draw the on-axis ray with the largest aperture that reaches the sensor. What acts as the aperture stop? Show in the drawing the entrance pupil and the exit pupil of the magnifying glass + objective system: where is the entrance pupil, is it real or virtual, and what is its diameter?

**6. Off-axis ray tracing.** For the case $e > 0$, draw the rays of full and limiting illumination, and determine at what distance from the centre of the paper the object points that mark these fields are located. Compare this with the photo from step 4.

**7. Discussion.** What would happen if we increased $e$? What role does the diameter of the magnifying glass play? And that of the aperture stop of the camera? Why is the edge of the magnifying glass not sharp in the photo? Apply the results to the use of a magnifying glass with the eye: where should the eye be placed, and why?
