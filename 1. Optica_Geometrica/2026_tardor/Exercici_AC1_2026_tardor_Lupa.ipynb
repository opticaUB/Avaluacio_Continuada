## **Exercici 1** d'Avaluació Continuada: Instruments Òptics

### El mòbil com a ull: estudi experimental de la lupa

<br>

Poseu el vostre **nom i cognoms**.

<br>

Des del punt de vista òptic, la càmera del mòbil és molt semblant a l'ull: té un sistema convergent de lents (l'objectiu, que fa de cristal·lí), un diafragma (la pupil·la), un sensor on es forma la imatge (la retina) i un sistema d'enfocament automàtic (l'acomodació). En aquest exercici fareu servir el mòbil com a **ull artificial** per estudiar experimentalment la **lupa**: quin augment visual proporciona, quina porció de l'objecte permet veure (el camp) i com es distribueix la il·luminació en la imatge, segons la distància de la lupa a l'ull.

L'exercici té dues parts: un petit **experiment** que fareu amb el vostre mòbil i una lupa, i el **desenvolupament teòric** que ha de permetre interpretar l'experiment.

#### Material

- El vostre mòbil (feu servir la càmera principal posterior).
- Una lupa, és a dir, una lent convergent, de focal entre 5 i 20 cm aproximadament. Serveix qualsevol lupa de basar o papereria (si no en trobeu cap, pregunteu al professor).
- Paper mil·limetrat (podeu imprimir l'adjunt a la tasca del CV).
- Un regle o una cinta mètrica.
- Opcional: una cartolina negra per definir millor el contorn de la lupa.
- Opcional: llibres o capses per fer de suport.

#### Preparació del mòbil i del muntatge

**Configuració de la càmera**

- Feu servir sempre la càmera principal (1x), sense zoom digital i sense flaix. Si el vostre mòbil té un mode macro automàtic, desactiveu-lo: molts mòbils canvien de càmera sense avisar quan l'objecte és a prop.
- En les darreres fotos cal que l'enfocament estigui **bloquejat a l'infinit**, com un ull relaxat que no acomoda.
    - a l'iPhone, enfoqueu un objecte llunyà i bloquegeu-ne l'enfocament mantenint el dit premut sobre la pantalla fins que aparegui «Bloqueig AE/AF».
    - a Android, feu servir el mode Pro o Manual (enfocament MF a ∞) o una aplicació de càmera amb enfocament manual (per exemple, Open Camera).
- Cada fotografia guarda unes metadades on hi ha, entre d'altres, la distància focal real de l'objectiu $f'_c$, el nombre de diafragma $N$, la focal equivalent en format de 35 mm i la càmera utilitzada. Les podeu consultar a la galeria del mòbil (detalls o informació de la foto) o a l'ordinador (a Windows: Propietats → Detalls).

**Contorn de la lupa**

Si la muntura no queda prou ben definida a l'última foto, podeu enganxar-hi una cartolina negra amb un forat circular centrat de diàmetre conegut, una mica més petit que la lupa. Per tant, el diàmetre efectiu de la lupa $D_L$ serà el del forat. 

**Muntatge suggerit**

El muntatge vertical sol ser el més còmode: el paper mil·limetrat sobre la taula i la lupa paral·lela al paper, aguantada a l'alçada adequada amb una pila de llibres. Així la lupa queda estable i teniu les mans lliures. El mòbil va a sobre, paral·lel a la lupa i amb la càmera centrada, aguantat amb una altra pila de llibres (o a mà) que us permeti variar la distància $e$ entre la lupa i el mòbil. Si queda molt fosc, procureu que arribi llum de costat al paper, perquè ni el mòbil ni la lupa hi facin ombra (mai amb el flaix de la càmera). Feu una foto del muntatge per incloure-la en el treball.

<br>


#### Guia de l'exercici

**1. Distància focal de la lupa.** Situeu-vos davant d'una paret que tingui una finestra a la paret oposada. Apropeu la lupa a la paret fins que s'hi formi la imatge nítida de la finestra (si es forma la imatge de l'exterior, millor encara). Estimeu la focal $f'_L$ de la lupa a partir de la distància lupa–paret. Feu una foto en què es vegi la lupa i la imatge de la finestra a la paret.

**2. Punt proper i foto de referència.** Sense la lupa, apropeu el vostre ull al paper mil·limetrat fins a trobar el vostre punt proper i mesureu-ne la distància $s_{pp}$. Poseu el mòbil a la mateixa distància del paper i feu una foto en què es vegi el paper mil·limetrat enquadrat i enfocat, amb l'enfocament automàtic activat (mode normal). Quin és el camp de visió de la càmera (compteu els quadrats del paper que hi caben a la fotografia)? Utilitzant les metadades, deduïu la mida del sensor i el diàmetre del diafragma d'obertura de la càmera $D_c$.

**3. Lupa enganxada a la càmera ($e = 0$).** Configureu la càmera perquè enfoqui a l'infinit i fixeu aquest paràmetre. Poseu la lupa centrada davant de la càmera, tan a prop com pugueu. Busqueu la distància lupa–paper a la qual el paper es veu enfocat i feu una foto.
- Quina distància hi ha entre el paper i la lupa? Per què?
- Comparant aquesta foto amb la del punt 2, obteniu l'augment visual que proporciona la lupa i compareu-lo amb el nominal de la lupa. Raoneu la resposta.
- Quin és ara el camp de visió? Què el limita?

**4. Separant la càmera de la lupa ($e > 0$).** Mantenint fixa la distància paper–lupa i l'enfocament de la càmera a l'infinit. Allunyeu la càmera de la lupa fins que la muntura de la lupa es vegi sencera a la foto. Feu una foto i mesureu la distància $e$ entre la càmera i la lupa.
- Segueix enfocat el paper mil·limetrat?
- Quin és l'augment ara?
- Quin és el camp de visió? Què el limita ara?
- Mesureu el radi de la zona amb il·luminació plena i el de la zona amb il·luminació límit comptant quadradets del paper mil·limetrat.

**5. Traçat de raigs en eix.** Per al cas $e > 0$, feu el traçat en eix del raig amb més obertura que arriba al sensor. Qui fa de diafragma d'obertura? Mostreu en el traçat la pupil·la d'entrada i la de sortida del sistema lupa + objectiu: on és la pupil·la d'entrada, és real o virtual i quin diàmetre té?

**6. Traçat de raigs en camp.** Per al cas $e > 0$, traceu els raigs d'il·luminació plena i límit, i determineu a quina distància del centre del paper es troben els punts objecte que marquen aquests camps. Compareu-ho amb la foto del punt 4.

**7. Discussió.** Què passaria si incrementéssim $e$? Quin paper hi juga el diàmetre de la lupa? I el del diafragma d'obertura de la càmera? Per què la vora de la lupa no surt nítida a la foto? Traslladeu els resultats a l'ús d'una lupa amb l'ull: on cal posar l'ull, i per què?
