A més, a continuació, teniu la solució que el professor ha fet per comprovar que tot fos correcte abans de publicar l'enunciat:

#### Dades i caracterització de la càmera

Totes les longituds en mm. Fotos preses amb un Motorola edge 60 pro (EXIF: $f'_c = 5{,}56$ mm, $N = 1{,}8$, focal equivalent 24 mm, imatges de 4096 × 3072 px). La focal equivalent dona directament la mida del sensor (factor de retall $c = 24/f'_c$) i, per tant, la relació entre píxels i angles: $k = (\text{píxels}) / (\text{mida del sensor}/f'_c)$, independent de $f'_c$.

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image, ImageOps
import io, os, urllib.request
from functools import lru_cache

# Les fotos de l'experiència es llegeixen del repositori de GitHub, així el quadern corre a Colab
# sense haver de pujar res. Si la foto és a la mateixa carpeta que el quadern, es fa servir la còpia local.
URL_FOTOS = "https://raw.githubusercontent.com/opticaUB/Avaluacio_Continuada/main/1.%20Optica_Geometrica/2026_tardor/"

@lru_cache(maxsize=None)
def _llegeix(fitxer):
    if os.path.exists(fitxer):
        with open(fitxer, "rb") as f:
            return f.read()
    with urllib.request.urlopen(URL_FOTOS + fitxer) as r:
        return r.read()

def foto(fitxer):
    """Obre una foto de l'experiència (local o de GitHub) amb l'orientació EXIF aplicada."""
    return ImageOps.exif_transpose(Image.open(io.BytesIO(_llegeix(fitxer))))

# --- Dades de l'experiència (mm) ---
f_L  = 150.0    # focal de la lupa (imatge de la finestra sobre la paret)
D_L  = 70.0     # diàmetre útil de la lupa
f_c  = 5.6      # focal de la càmera (EXIF: 5.56 mm)
N    = 1.8      # nombre de diafragma (EXIF)
D    = 3.0      # diafragma d'obertura de la càmera (f_c/N = 3.1 mm)
s_pp = 250.0    # punt proper de l'ull (i distància paper-mòbil de la foto de referència)
f35  = 24.0     # focal equivalent en 35 mm (EXIF)
Npx  = (4096, 3072)   # píxels (costat llarg, costat curt)

# Sensor a partir del factor de retall (format 4:3, diagonal del 35 mm = 43.27 mm)
diag  = 43.27 * f_c / f35
W, H  = diag * 4/5, diag * 3/5      # mm
pitch = W / Npx[0]                  # mida del píxel
k     = f_c / pitch                 # píxels per radiant (tan) al sensor
tan_c = np.array([W, H]) / 2 / f_c  # tangent del semicamp de la càmera (llarg, curt)

print(f"sensor = {W:.2f} x {H:.2f} mm (diagonal {diag:.2f} mm), píxel = {pitch*1e3:.2f} µm")
print(f"k = {k:.0f} px/rad,  semicamp de la càmera: {np.degrees(np.arctan(tan_c)).round(1)} graus")
print(f"D = f_c/N = {f_c/N:.2f} mm  (fem servir D = {D} mm)")
```

#### 1. Focal de la lupa

La imatge de la finestra (o de l'exterior) es forma a la paret amb la lupa a $\approx 150$ mm: $f'_L \approx 150$ mm. Com que l'objecte és a una distància finita $L$, la imatge es forma a $s' = f'_L\,L/(L-f'_L) \approx f'_L\,(1 + f'_L/L)$: **sobreestimem** la focal en un factor $f'_L/L$, un 3–5 % amb la finestra a 3–5 m, i molt menys si la imatge nítida és la de l'exterior.

L'augment nominal (ull estàndard, $s_{pp} = 250$ mm, imatge a l'infinit) és $\Gamma = 250/f'_L \approx 1{,}67$.

```python
plt.imshow(foto("estimacio-focal-lupa.jpg"))
plt.axis("off");
```

#### 2. Punt proper i foto de referència

Amb el mòbil a $s_{pp} = 250$ mm del paper, un mil·límetre del paper ocupa $k/s_{pp}$ píxels i el camp de visió és $s_{pp}\cdot W/f'_c \times s_{pp}\cdot H/f'_c$. Sobre la foto, el pas de les línies de 10 mm és de ≈ 109,5 px, és a dir, **10,95 px/mm**.

```python
pmm_ref = 10.95                       # px/mm mesurats a la foto de referència (s = 250 mm)
camp_pred = s_pp * np.array([W, H]) / f_c
camp_mes  = np.array(Npx) / pmm_ref
print(f"Camp de visió a s_pp: previst {camp_pred.round(0)} mm, mesurat {camp_mes.round(0)} mm")
print(f"Distància efectiva paper-PE de la càmera: k/pmm = {k/pmm_ref:.0f} mm "
      f"(la pupil·la d'entrada és uns mil·límetres dins del mòbil)")
```

#### 3–4. Augment visual i camp: les fotos

Amb el paper al pla focal objecte de la lupa, cada punt del paper a alçada $y$ surt de la lupa com un **feix de raigs paral·lels** amb $\tan\omega' = y/f'_L$. Per això, amb la càmera enfocada a l'infinit, el paper es veu nítid **sigui quina sigui $e$** (punts 3 i 4), i la mida de la imatge al sensor, $k\,y/f'_L$, **no depèn de $e$**. Sense lupa, a la distància del punt proper, $\tan\omega = y/s_{pp}$. L'augment visual és

$$\Gamma = \frac{\tan\omega'}{\tan\omega} = \frac{s_{pp}}{f'_L},$$

que per a l'ull (i per a la càmera, que hem posat a $s_{pp}$) val $250/150 \approx 1{,}67$. Mesurem el pas de la quadrícula al centre de cada foto (línies de 10 mm):

```python
fotos = {   # nom: (fitxer, e [mm], px/mm mesurats al centre)
    "s = 250 mm (sense lupa)": ("sense-lupa-s250mm.jpg", None, 10.95),
    "e = 0":                   ("amb-lupa-e0.jpg",         0.0, 18.0),
    "e = 80 mm":               ("amb-lupa-e80mm.jpg",     80.0, 17.5),
}
for nom, (fit, e, pmm) in fotos.items():
    if e is None: continue
    print(f"{nom:10s}  {pmm:5.1f} px/mm   Gamma = {pmm/pmm_ref:.2f}")
pm = np.mean([v[2] for v in fotos.values() if v[1] is not None])
print(f"\nGamma mitjà = {pm/pmm_ref:.2f}   (nominal s_pp/f'_L = {s_pp/f_L:.2f})")
print(f"Focal efectiva de la lupa a partir de k: f'_L = k/pmm = {k/pm:.0f} mm")

fig, axs = plt.subplots(1, 3, figsize=(13, 5.8))
for ax, (nom, (fit, e, pmm)) in zip(axs, fotos.items()):
    im = foto(fit); im.thumbnail((600, 800))
    ax.imshow(im); ax.set_title(nom, fontsize=10); ax.axis("off")
plt.tight_layout(); plt.show()
```

L'augment és el mateix en les dues fotos amb lupa (dins de l'error de la mesura) i val $\Gamma \approx 1{,}62$, un 3 % per sota del nominal $1{,}67$. La comparació és independent de la calibració de la càmera ($k$ es cancel·la en el quocient). Si es fa servir $k$, surten valors un 4–7 % més grans tant per a $f'_L$ com per a la distància de la foto de referència: la focal mesurada a la paret és una estimació aproximada i les distàncies es mesuren des del vidre del mòbil, no des de la seva pupil·la.

El que sí que canvia amb $e$ és **el camp**: amb $e = 0$ la imatge omple tot el sensor; en allunyar la càmera apareix el contorn de la lupa, que tanca una zona més petita del paper.

#### 5–6. Model: diafragmes, pupil·les i camps

Eix òptic d'esquerra a dreta (sentit de la llum) i origen a la lupa: paper a $x = -f'_L$, lupa a $x = 0$ i diafragma de la càmera a $x = e$.

- **Diafragma d'obertura (DO):** el de la càmera, de diàmetre $D \approx 3$ mm $\ll D_L$. El raig en eix amb més obertura surt del centre del paper, travessa la lupa a una alçada $D/2$ i, ja paral·lel a l'eix, arriba a la vora del DO.
- **Pupil·la de sortida (PS):** la de la càmera (a l'espai imatge no hi ha res més després del DO).
- **Pupil·la d'entrada (PE):** la imatge del DO a través dels elements que té al davant, és a dir, de la lupa. Amb l'equació de Gauss, $\dfrac{1}{s'} - \dfrac{1}{s} = \dfrac{1}{f'_L}$, i posant-hi $s' = e$ (més avall s'explica per què):
$$ s_{PE} = s = \frac{e\, f'_L}{f'_L - e}, \qquad D_{PE} = D\,\frac{f'_L}{f'_L - e}.$$
  - Amb $e = 0$ coincideix amb el DO.
  - Amb $0 < e < f'_L$, $s_{PE} > 0$: és **virtual**, a la dreta de la lupa i més enllà de la càmera ($s_{PE} > e$), i més gran que el DO.
  - Amb $e > f'_L$, $s_{PE} < 0$: és **real**, a l'esquerra de la lupa i més enllà del paper.
  - Amb $e = f'_L$ és a l'infinit.
- **Diafragma de camp:** la muntura de la lupa (la lluerna d'entrada coincideix amb ella mateixa, perquè és a la lupa), llevat que el sensor talli abans: amb $e = 0$ el camp el limita el sensor, $y = f'_L \tan\omega'_c$.

**Per què $s' = e$ i no $s = e$.** És el punt clau de l'exercici.

La PE és la imatge del DO *vista des de l'espai objecte*: el punt cap on apunten, abans d'entrar a la lupa, els raigs que després passaran per la vora del DO. El DO, però, és físicament **darrere** de la lupa, a la zona per on la llum ja ha sortit. En el sentit real de la llum (d'esquerra a dreta), doncs, el DO és a l'**espai imatge** de la lupa: la seva posició és una distància imatge, $s' = e$, i la incògnita és el punt conjugat a l'espai objecte, $s = s_{PE}$. No cal inventar cap objecte nou: la pregunta és quin punt de l'espai objecte conjuga la lupa amb el DO.

Amb $s' = e$:
$$\frac{1}{e} - \frac{1}{s} = \frac{1}{f'_L} \quad\Longrightarrow\quad s = \frac{e\,f'_L}{f'_L - e}.$$
La mida surt de l'augment lateral, $\beta' = s'/s = (f'_L - e)/f'_L$. Com que el DO fa el paper d'imatge, $D = \beta' D_{PE}$ i, per tant, $D_{PE} = D\,f'_L/(f'_L - e)$.

El signe es llegeix amb el conveni habitual per a l'espai objecte. Si $s < 0$ (a l'esquerra), la PE és real: els raigs hi passen de debò abans d'arribar a la lupa. Si $s > 0$ (a la dreta), la PE és virtual: els raigs de l'espai objecte hi apunten, però la lupa els desvia abans d'arribar-hi. Per això als traçats la PE s'assenyala amb prolongacions discontínues.

**Alternativa: invertir el sentit de la llum.** També es pot imaginar que la llum surt del DO i viatja cap a l'esquerra. Llavors el DO és un objecte real situat davant de la lupa i, amb l'eix orientat en el nou sentit de propagació, la seva distància és $s = -e$. Amb la mateixa equació i la mateixa $f'_L$ (una lent prima convergent ho és en tots dos sentits):
$$\frac{1}{s'} + \frac{1}{e} = \frac{1}{f'_L} \quad\Longrightarrow\quad s' = \frac{e\,f'_L}{e - f'_L} = -\frac{e\,f'_L}{f'_L - e}.$$
Amb $e < f'_L$ surt $s' < 0$: la imatge és al mateix costat que l'objecte (el DO), és a dir, és virtual i a la dreta de la lupa. És el mateix punt que abans; el signe és el contrari només perquè ara l'eix apunta cap a l'altra banda. Qui segueixi aquest camí ha de recordar, en acabar, que les distàncies estan mesurades en sentit invers.

**L'error que cal evitar** és barrejar els dos camins: posar el DO com a objecte i mantenir la llum d'esquerra a dreta.
- Amb $s = +e$ surt $s' = e\,f'_L/(f'_L + e)$, que no és la PE: és la imatge d'un objecte virtual situat al DO, una cosa que no passa en aquest sistema.
- Amb $s = -e$ el càlcul és el mateix que el de l'alternativa i surt la distància correcta, $e\,f'_L/(f'_L - e)$. L'error és llegir després el signe de $s'$ amb el conveni del sentit original («$s' < 0$, a l'esquerra»), quan $s = -e$ només té sentit amb la llum invertida. Així la PE acaba al costat equivocat de la lupa.

**Camps.** El feix que surt del punt objecte a alçada $y$ és paral·lel, amb $\tan\omega' = y/f'_L$; el seu raig principal passa pel centre del DO i talla la lupa a l'alçada $e\,y/f'_L$. El DO en selecciona una «taca» de diàmetre $D$ centrada en aquest punt. Per tant:

$$ y_\text{plena} = \frac{f'_L\,(D_L - D)}{2e}, \qquad y_\text{límit} = \frac{f'_L\,(D_L + D)}{2e}. $$

El mateix resultat surt traçant des de la PE: la recta que va del centre de la PE a la vora de la lupa talla el paper a $y_\text{mitja}$. L'amplada de la transició, $f'_L D/e$ sobre el paper (angle $D/e$), és exactament el cercle de desenfocament de la vora de la lupa, que és a una distància $e$ mentre la càmera enfoca a l'infinit: **per això la vora no surt nítida**.

La muntura (de diàmetre exterior $D_m \geq D_L$) es veu sencera a la foto quan $D_m/2e$ és menor que el semicamp de la càmera en la direcció curta del sensor, és a dir, per a $e \gtrsim D_m/(2\tan\omega'_{c})$.

```python
R = D_L / 2
casos = [0.0, 80.0]
print(f"La vora útil de la lupa entra sencera al fotograma per a e > (D_L+D)/(2 tan w'_c,curt) = {(D_L+D)/(2*tan_c[1]):.0f} mm\n")
print(f"{'e':>6} | {'s_PE':>8} {'D_PE':>6} | {'y_plena':>8} {'y_mitja':>8} {'y_límit':>8} (mm al paper) | r_mitja (px)")
for e in casos:
    if e == 0:
        print(f"{e:6.0f} | {0:8.1f} {D:6.2f} | camp limitat pel sensor: {f_L*tan_c[1]:.0f} x {f_L*tan_c[0]:.0f} mm (semieixos)")
        continue
    sPE, DPE = e*f_L/(f_L-e), D*f_L/(f_L-e)
    yp, ym, yl = f_L*(D_L-D)/(2*e), f_L*D_L/(2*e), f_L*(D_L+D)/(2*e)
    print(f"{e:6.0f} | {sPE:8.1f} {DPE:6.2f} | {yp:8.1f} {ym:8.1f} {yl:8.1f}               | {k*R/e:7.0f}")
```

```python
e, fit, cx, cy, r_mes = 80.0, "amb-lupa-e80mm.jpg", 1165, 2162, 1223   # centre i radi de la vora ajustats a la foto vertical (px)
pmm_e = 17.5                                                             # px/mm al centre de la foto
fig, ax = plt.subplots(figsize=(7.8, 10))
t = np.linspace(0, 2*np.pi, 400)
im = foto(fit); s = 4
im = im.resize((im.width//s, im.height//s)); ax.imshow(im)
for DD, c, ls in ((D_L-D, "lime", "-"), (D_L, "yellow", "--"), (D_L+D, "red", "-")):
    r = k*DD/(2*e)/s
    ax.plot(cx/s + r*np.cos(t), cy/s + r*np.sin(t), color=c, ls=ls, lw=1.2)
ax.plot(cx/s + r_mes/s*np.cos(t), cy/s + r_mes/s*np.sin(t), color="cyan", ls=":", lw=2)
ax.set_xlim(0, im.width); ax.set_ylim(im.height, 0); ax.axis("off"); ax.set_title(f"e = {e:.0f} mm")
plt.tight_layout(); plt.show()

print(f"r_mitja previst {k*D_L/(2*e):.0f} px, mesurat {r_mes} px")
print(f"camp (radi mitja sobre el paper): previst {f_L*D_L/(2*e):.0f} mm, mesurat {r_mes/pmm_e:.0f} mm")
print(f"diàmetre útil de la lupa que reprodueix la foto: 2 e r / k = {2*e*r_mes/k:.1f} mm  (D_L = {D_L:.0f} mm)")

# Amplada de la transició (10 %-90 % de la caiguda) mesurada al llarg de la vora: mediana 73 px (P25-P75: 69-80 px).
# Per a un disc (el DO) que creua una vora recta, el 10-90 % de l'àrea ocupa 0,687 D.
dr = 73
print(f"amplada de la transició: prevista {0.687*k*D/e:.0f} px, mesurada {dr} px  ->  D = {dr/0.687/k*e:.1f} mm  (EXIF: f_c/N = {f_c/N:.1f} mm)")
```

El radi del camp mesurat directament sobre el paper (70 mm) és un 6 % més gran que el previst (66 mm), la mateixa diferència que hem trobat entre la focal efectiva de la lupa ($k/\text{pmm} \approx 160$ mm) i l'estimada a la paret (150 mm). #### 7. Discussió

- **En augmentar $e$**, l'augment no canvia (el feix és paral·lel), però el camp es redueix com $1/e$: per veure el màxim de paper cal posar l'ull **tan a prop de la lupa com es pugui**. Amb l'ull, els raigs principals passen pel centre de la pupil·la de l'ull, que fa de DO.
- **El diàmetre de la lupa** fixa el camp (proporcional a $D_L$), no l'augment, que només depèn de $f'_L$. Una lupa més gran no augmenta més, però deixa veure més.
- **El diàmetre del DO** (la pupil·la de l'ull, de 2 a 8 mm segons la llum) només fixa l'amplada de la transició entre la il·luminació plena i la límit, $f'_L D/e$ sobre el paper. Amb molta llum (pupil·la petita) la vora del camp es veu més definida.
- **La vora de la lupa no surt nítida** perquè és a una distància $e$ de la càmera, que enfoca a l'infinit: el seu cercle de desenfocament, d'angle $D/e$, és precisament la zona de transició entre la il·luminació plena i la límit.
